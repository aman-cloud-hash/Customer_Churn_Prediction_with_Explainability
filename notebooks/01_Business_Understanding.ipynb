# Customer Churn Prediction with Explainability

## Business Understanding

**Project Type:** Data Analytics + Machine Learning

**Dataset:** IBM Telco Customer Churn Dataset

**Objective:**
Analyze customer churn, identify the key reasons customers leave, build an interpretable machine learning model, and provide actionable business recommendations to improve customer retention.

## Business Background

Customer churn is one of the biggest challenges faced by telecom companies.

When a customer stops using the company's services, it is called customer churn.

Losing customers reduces revenue and increases customer acquisition costs because acquiring a new customer is usually more expensive than retaining an existing one.

By analyzing customer behavior and predicting churn in advance, the company can take preventive actions such as offering discounts, improving customer support, or providing better plans to retain valuable customers.

## Problem Statement

The telecom company is experiencing customer churn, leading to revenue loss and increased customer acquisition costs.

The company wants to identify the major factors responsible for customer churn and predict which customers are likely to leave.

The prediction model should also be explainable so that business stakeholders can understand why a customer is predicted to churn.

## Business Objectives

- Analyze customer demographics and service usage.
- Identify the major factors influencing churn.
- Calculate important business KPIs.
- Build an interpretable Logistic Regression model.
- Explain predictions using SHAP.
- Create an interactive Streamlit dashboard.
- Provide business recommendations to reduce churn.

## Expected Business Impact

If implemented successfully, this project can help the company:

- Reduce customer churn.
- Increase customer retention.
- Improve customer satisfaction.
- Increase long-term revenue.
- Support data-driven business decisions.
