# Business Recommendations

## Objective

Based on exploratory data analysis and machine learning results, provide actionable business recommendations to reduce customer churn.

These recommendations are intended for business stakeholders, marketing teams, and management.

```python
import pandas as pd
```

```python
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

df.head()
```

```python
churn_rate = (df["Churn"] == "Yes").mean() * 100

print(f"Overall Churn Rate : {churn_rate:.2f}%")
```

```python
contract = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

contract
```

```python
internet = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

internet
```

```python
payment = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

payment
```

```python
senior = pd.crosstab(
    df["SeniorCitizen"],
    df["Churn"],
    normalize="index"
) * 100

senior
```

```python
monthly = df.groupby("Churn")["MonthlyCharges"].mean()

monthly
```

```python
tenure = df.groupby("Churn")["tenure"].mean()

tenure
```

```python
recommendations = pd.DataFrame({

    "Problem":[

        "Month-to-Month Customers",

        "High Monthly Charges",

        "Fiber Customers",

        "Electronic Check Users",

        "Low Tenure Customers"

    ],

    "Recommendation":[

        "Offer Long-Term Contracts",

        "Provide Discounts",

        "Improve Network Quality",

        "Encourage Auto Payment",

        "Onboarding Campaign"

    ]

})

recommendations
```

```python
recommendations.to_csv(
    "../data/business_recommendations.csv",
    index=False
)

print("Recommendations Saved!")
```

# Recommendation 1

Encourage customers to move from Month-to-Month contracts to yearly plans.

Reason:
Long-term customers have significantly lower churn.

# Recommendation 2

Provide loyalty rewards after 6 months and 12 months.

Reason:
Customers with higher tenure rarely churn.

# Recommendation 3

Offer personalized discounts for customers with high monthly charges.

Reason:
Expensive plans increase churn risk.

# Recommendation 4

Improve Fiber Optic service quality.

Reason:
Fiber customers show higher churn than DSL customers.

# Recommendation 5

Promote AutoPay and Bank Transfer payment methods.

Reason:
Electronic Check users churn more frequently.

# Recommendation 6

Identify high-risk customers using the prediction model and contact them proactively before they leave.
