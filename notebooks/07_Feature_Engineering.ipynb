# Feature Engineering

## Objective

In this notebook we prepare the cleaned dataset for Machine Learning.

Steps:

- Feature Selection
- Encoding
- Feature Scaling
- Train-Test Split
- Save processed data

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler

import os

os.makedirs("../models", exist_ok=True)

import pickle
```

```python
df = pd.read_csv("../data/clean_telco.csv")
```

```python
df.head()
```

```python
df.info()
```

## Separate Input and Target Variables

Target Variable:

- Churn

Input Variables:

- Remaining columns

```python
X = df.drop("Churn", axis=1)
y = df["Churn"]
```

```python
print(X.shape)
print(y.shape)
```

```python
label_encoder = LabelEncoder()

y = label_encoder.fit_transform(y)
```

```python
pd.Series(y).value_counts()
```

```python
categorical_columns = X.select_dtypes(include="object").columns

categorical_columns
```

```python
X = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True
)
```

```python
X.head()
```

```python
X.shape
```

## Train Test Split

Training Data = 80%

Testing Data = 20%

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
scaler = StandardScaler()
```

```python
numerical_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]
```

```python
X_train[numerical_columns] = scaler.fit_transform(
    X_train[numerical_columns]
)
```

```python
X_test[numerical_columns] = scaler.transform(
    X_test[numerical_columns]
)
```

```python
X_train.head()
```

```python
with open("../models/scaler.pkl","wb") as file:
    pickle.dump(scaler,file)

print("Scaler Saved!")
```

```python
with open("../models/label_encoder.pkl","wb") as file:
    pickle.dump(label_encoder,file)

print("Label Encoder Saved!")
```

```python
X_train.to_csv("../data/X_train.csv", index=False)
X_test.to_csv("../data/X_test.csv", index=False)

pd.Series(y_train).to_csv("../data/y_train.csv", index=False)
pd.Series(y_test).to_csv("../data/y_test.csv", index=False)
```

```python
print("Processed datasets saved successfully!")
```

```python
feature_names = X.columns

feature_names
```

```python
len(feature_names)
```

```python
with open("../models/feature_names.pkl","wb") as file:
    pickle.dump(feature_names,file)

print("Feature Names Saved!")
```
